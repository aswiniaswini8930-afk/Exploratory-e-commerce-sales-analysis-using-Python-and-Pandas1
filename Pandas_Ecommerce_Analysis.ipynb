```python
# Keep an untouched copy of the original dataset
df_raw = df.copy(deep=True)

# Create a separate DataFrame for our work
df_work = df_raw.copy(deep=True)

# Verify their dimensions
print("Original df shape:", df.shape)
print("Raw backup shape:", df_raw.shape)
print("Working copy shape:", df_work.shape)
```

```python
df_work.rename(
    columns={"Description": "ProductDescription"},
    inplace=True
)

df_work.columns
```

```python
# Calculate revenue for each transaction line
df_work["Revenue"] = df_work["Quantity"] * df_work["UnitPrice"]

# Display the calculation for the first 10 rows
df_work[["Quantity", "UnitPrice", "Revenue"]].head(10)
```

```python
# Identify cancellation invoices
is_cancellation = (
    df_work["InvoiceNo"]
    .astype("string")
    .str.startswith("C", na=False)
)

# Give every row a default category
df_work["TransactionType"] = "Other/Invalid"

# Classify cancellations first
df_work.loc[
    is_cancellation,
    "TransactionType"
] = "Cancellation"

# Classify negative quantities outside cancellation invoices
df_work.loc[
    (~is_cancellation) & (df_work["Quantity"] < 0),
    "TransactionType"
] = "Return"

# Classify positive quantities with positive prices
df_work.loc[
    (~is_cancellation)
    & (df_work["Quantity"] > 0)
    & (df_work["UnitPrice"] > 0),
    "TransactionType"
] = "Positive Sale"

# Inspect the classifications
print(df_work[
    ["InvoiceNo", "Quantity", "UnitPrice", "Revenue", "TransactionType"]
].head(10))

print("\nTransaction type counts:")
print(df_work["TransactionType"].value_counts())
```

```python
df_work["IsIdentifiedCustomer"] = df_work["CustomerID"].notna()

df_work[["CustomerID", "IsIdentifiedCustomer"]].head(10)
```

```python
# Convert InvoiceDate to Pandas datetime
df_work["InvoiceDate"] = pd.to_datetime(df_work["InvoiceDate"])

# Verify the data type
print(df_work["InvoiceDate"].dtype)

# Display a few dates
df_work["InvoiceDate"].head()
```

```python
# Convert identifier and country columns to string
df_work["InvoiceNo"] = df_work["InvoiceNo"].astype("string")
df_work["StockCode"] = df_work["StockCode"].astype("string")
df_work["Country"] = df_work["Country"].astype("string")

# Verify the data types
print(df_work[["InvoiceNo", "StockCode", "Country"]].dtypes)

# Display the first five rows
df_work[["InvoiceNo", "StockCode", "Country"]].head()
```

```python
# Define the desired column order
column_order = [
    "InvoiceNo",
    "InvoiceDate",
    "StockCode",
    "ProductDescription",
    "CustomerID",
    "IsIdentifiedCustomer",
    "Country",
    "Quantity",
    "UnitPrice",
    "Revenue",
    "TransactionType"
]

# Reorder the columns in the working DataFrame
df_work = df_work[column_order]

# Verify the new order
print(df_work.columns.tolist())

# Display the first five rows
df_work.head()
```

```python
import pandas as pd

df = pd.read_excel(
    "/content/drive/MyDrive/Online Retail.xlsx"
)

df_raw = df.copy(deep=True)
df_work = df_raw.copy(deep=True)

print("Original dataset:", df.shape)
print("Raw backup:", df_raw.shape)
print("Working copy:", df_work.shape)
```

```python
df_work.rename(
    columns={"Description": "ProductDescription"},
    inplace=True
)

print(df_work.columns.tolist())
```

```python
df_work["Revenue"] = df_work["Quantity"] * df_work["UnitPrice"]

df_work[["Quantity", "UnitPrice", "Revenue"]].head(10)
```

```python
# Rule 1: Invoice numbers starting with C are cancellations
is_cancellation = (
    df_work["InvoiceNo"].astype("string").str.startswith("C", na=False)
)

df_work["TransactionType"] = "Other/Invalid"

# Rule 2: Cancellation
df_work.loc[is_cancellation, "TransactionType"] = "Cancellation"

# Rule 3: Negative quantity, excluding cancellations
df_work.loc[
    (~is_cancellation) & (df_work["Quantity"] < 0),
    "TransactionType"
] = "Return"

# Rule 4: Positive quantity and positive price, excluding cancellations
df_work.loc[
    (~is_cancellation)
    & (df_work["Quantity"] > 0)
    & (df_work["UnitPrice"] > 0),
    "TransactionType"
] = "Positive Sale"

# Display sample records
print(df_work[[
    "InvoiceNo", "Quantity", "UnitPrice",
    "Revenue", "TransactionType"
]].head(10))

print("\nTransaction type counts:")
print(df_work["TransactionType"].value_counts())
```

```python
df_work["IsIdentifiedCustomer"] = df_work["CustomerID"].notna()

df_work[[
    "CustomerID",
    "IsIdentifiedCustomer"
]].head(10)
```

```python
df_work["InvoiceDate"] = pd.to_datetime(
    df_work["InvoiceDate"]
)

print(df_work["InvoiceDate"].dtype)
print(df_work["InvoiceDate"].head())
```

```python
df_work["InvoiceNo"] = df_work["InvoiceNo"].astype("string")
df_work["StockCode"] = df_work["StockCode"].astype("string")
df_work["Country"] = df_work["Country"].astype("string")

print(df_work[["InvoiceNo", "StockCode", "Country"]].dtypes)
```

```python
column_order = [
    "InvoiceNo",
    "InvoiceDate",
    "StockCode",
    "ProductDescription",
    "CustomerID",
    "IsIdentifiedCustomer",
    "Country",
    "Quantity",
    "UnitPrice",
    "Revenue",
    "TransactionType"
]

df_work = df_work[column_order]

print(df_work.columns.tolist())
df_work.head()
```

```python
df_work = df_work.sort_values(
    by=["InvoiceDate", "InvoiceNo"]
)

df_work.head()
```

```python
df_work.reset_index(drop=True, inplace=True)

print(df_work.head())

print("\nIs the index sequential?")
print(df_work.index.equals(pd.RangeIndex(len(df_work))))

print("\nTotal rows:", len(df_work))
print("First index:", df_work.index[0])
print("Last index:", df_work.index[-1])
```

```python
missing_count = df_work.isnull().sum()

print(missing_count)
```

```python
missing_count = df_work.isnull().sum()

missing_percentage = (
    missing_count / len(df_work)
) * 100

print(missing_percentage.sort_values(ascending=False))
```

```python
missing_report = pd.DataFrame({
    "MissingCount": df_work.isnull().sum(),
    "NonMissingCount": df_work.notnull().sum(),
    "MissingPercentage": (
        df_work.isnull().sum() / len(df_work)
    ) * 100
})

missing_report = missing_report.sort_values(
    by="MissingPercentage",
    ascending=False
)

missing_report
```

```python
missing_customer_rows = df_work[
    df_work["CustomerID"].isnull()
]

print("Rows with missing CustomerID:",
      len(missing_customer_rows))

missing_customer_rows.head(10)
```

```python
description = df_work["ProductDescription"]

missing_description_rows = df_work[
    description.isna()
    | description.astype("string").str.strip().eq("").fillna(False)
]

print("Rows with missing or blank descriptions:",
      len(missing_description_rows))

missing_description_rows.head(10)
```

```python
duplicate_count = df_work.duplicated().sum()

print("Exact duplicate rows before removal:",
      duplicate_count)
```

```python
rows_before = len(df_work)

df_work = df_work.drop_duplicates()

rows_after = len(df_work)

print("Rows before removal:", rows_before)
print("Duplicate rows removed:", rows_before - rows_after)
print("Rows after removal:", rows_after)

print("Remaining exact duplicates:",
      df_work.duplicated().sum())
```

```python
combo_counts = (
    df_work.groupby(["InvoiceNo", "StockCode"])
    .size()
    .reset_index(name="LineCount")
)

repeated_combinations = combo_counts[
    combo_counts["LineCount"] > 1
]

print(
    "Repeated invoice-product combinations:",
    len(repeated_combinations)
)

repeated_combinations.head(10)
```

```python
missing_before = df_work["ProductDescription"].isna().sum()

print("Missing descriptions before handling:", missing_before)
```

```python
df_work["ProductDescription"] = (
    df_work["ProductDescription"].fillna("Unknown Product")
)

print(
    "Missing descriptions after handling:",
    df_work["ProductDescription"].isna().sum()
)

print(
    "Unknown Product placeholders:",
    (df_work["ProductDescription"] == "Unknown Product").sum()
)
```

```python
quality_comparison = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Missing values across all columns",
        "Exact duplicate rows"
    ],
    "Before cleaning": [
        len(df_raw),
        int(df_raw.isna().sum().sum()),
        int(df_raw.duplicated().sum())
    ],
    "After cleaning": [
        len(df_work),
        int(df_work.isna().sum().sum()),
        int(df_work.duplicated().sum())
    ]
})

quality_comparison
```
